```python
# ==========================
# Experiment 7
# AI Code Generation using CodeGen
# ==========================


!pip -q install transformers accelerate


from transformers import pipeline


generator = pipeline(
    "text-generation",
    model="Salesforce/codegen-350M-mono"
)


prompt = """
# Python function to check whether a number is prime
"""


result = generator(
    prompt,
    max_new_tokens=150,
    do_sample=False,
    pad_token_id=generator.tokenizer.eos_token_id
)


print(result[0]["generated_text"])
```
