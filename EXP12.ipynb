```python
import gradio as gr
from transformers import pipeline

# Load GPT-2 model
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# Function to generate text
def generate_text(prompt):
    result = generator(
        prompt,
        max_new_tokens=60,
        do_sample=True,
        temperature=0.8,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    return result[0]["generated_text"]

# Create Gradio Interface
demo = gr.Interface(
    fn=generate_text,
    inputs=gr.Textbox(
        lines=3,
        placeholder="Enter your prompt here...",
        label="Input Prompt"
    ),
    outputs=gr.Textbox(
        lines=8,
        label="Generated Text"
    ),
    title="Generative AI Text Generator",
    description="Generate text using GPT-2 with Hugging Face and Gradio."
)

# Launch App
demo.launch(share=True, debug=True)
```
